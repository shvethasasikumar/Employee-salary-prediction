```python
# ============================================================
# EXPERIMENT 4
# EXPLORATORY DATA ANALYSIS (EDA)
# DATA INSPECTION, FILTERING, STATISTICAL ANALYSIS
# AND CORRELATION
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import zipfile
import glob

from google.colab import files


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("Please upload Employee_Salary_Dataset.csv or archive.zip")

uploaded = files.upload()

uploaded_file = list(uploaded.keys())[0]

print("\nUploaded file:", uploaded_file)


# ============================================================
# 3. FIND CSV FILE
# ============================================================

if uploaded_file.lower().endswith(".zip"):

    print("\nZIP file detected. Extracting...")

    with zipfile.ZipFile(uploaded_file, "r") as zip_ref:
        zip_ref.extractall("extracted_data")

    csv_files = glob.glob(
        "extracted_data/**/*.csv",
        recursive=True
    )

    if len(csv_files) == 0:
        raise FileNotFoundError(
            "No CSV file found inside the ZIP file."
        )

    csv_file = csv_files[0]

    print("CSV file found:", csv_file)

elif uploaded_file.lower().endswith(".csv"):

    csv_file = uploaded_file

else:

    raise ValueError(
        "Please upload a CSV or ZIP file."
    )


# ============================================================
# 4. LOAD DATASET
# ============================================================

df = pd.read_csv(csv_file)

print("\n================ DATASET LOADED ================")

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 5. DATASET INSPECTION
# ============================================================

print("\n================ DATASET INSPECTION ================")

# Display complete dataset
print("\nComplete Dataset:")
print(df)

# Dataset dimensions
print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types
print("\nData Types:")
print(df.dtypes)

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())


# ============================================================
# 6. FILTERING DATA
# ============================================================

print("\n================ DATA FILTERING ================")


# Employees with salary greater than 50000

high_salary = df[
    df["Salary"] > 50000
]

print("\nEmployees with Salary greater than 50000:")

print(high_salary)


# Employees with more than 5 years experience

experienced_employees = df[
    df["Experience_Years"] > 5
]

print("\nEmployees with more than 5 years of experience:")

print(experienced_employees)


# Employees older than 30

older_employees = df[
    df["Age"] > 30
]

print("\nEmployees older than 30:")

print(older_employees)


# ============================================================
# 7. SORTING DATA
# ============================================================

print("\n================ SORTING DATA ================")


# Sort salary in descending order

sorted_salary = df.sort_values(
    by="Salary",
    ascending=False
)

print("\nEmployees sorted by Salary (Descending):")

print(
    sorted_salary[
        [
            "ID",
            "Age",
            "Experience_Years",
            "Salary"
        ]
    ].head(10)
)


# Sort experience in descending order

sorted_experience = df.sort_values(
    by="Experience_Years",
    ascending=False
)

print("\nEmployees sorted by Experience (Descending):")

print(
    sorted_experience[
        [
            "ID",
            "Experience_Years",
            "Salary"
        ]
    ].head(10)
)


# ============================================================
# 8. SALARY RANKING VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

top_employees = sorted_salary.head(10)

plt.bar(
    top_employees["ID"].astype(str),
    top_employees["Salary"]
)

plt.xlabel("Employee ID")

plt.ylabel("Salary")

plt.title(
    "Employee Salary Ranking"
)

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 9. STATISTICAL ANALYSIS
# ============================================================

print("\n================ STATISTICAL ANALYSIS ================")


# Salary statistics

salary_mean = df["Salary"].mean()

salary_median = df["Salary"].median()

salary_mode = df["Salary"].mode()

salary_std = df["Salary"].std()


print("\nSalary Statistics:")

print(
    "Mean Salary:",
    round(salary_mean, 2)
)

print(
    "Median Salary:",
    round(salary_median, 2)
)

print(
    "Mode Salary:",
    salary_mode.tolist()
)

print(
    "Standard Deviation:",
    round(salary_std, 2)
)


# Experience statistics

print("\nExperience Statistics:")

print(
    "Mean Experience:",
    round(
        df["Experience_Years"].mean(),
        2
    )
)

print(
    "Median Experience:",
    round(
        df["Experience_Years"].median(),
        2
    )
)

print(
    "Standard Deviation:",
    round(
        df["Experience_Years"].std(),
        2
    )
)


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\n================ DESCRIPTIVE STATISTICS ================")

print(
    df[
        [
            "Age",
            "Experience_Years",
            "Salary"
        ]
    ].describe()
)


# ============================================================
# 11. GROUPING AND AGGREGATION
# ============================================================

print("\n================ GROUPING AND AGGREGATION ================")


# Average salary by gender

gender_avg_salary = df.groupby(
    "Gender"
)["Salary"].mean()


print("\nAverage Salary by Gender:")

print(gender_avg_salary)


# Average experience by gender

gender_avg_experience = df.groupby(
    "Gender"
)["Experience_Years"].mean()


print("\nAverage Experience by Gender:")

print(gender_avg_experience)


# Average age by gender

gender_avg_age = df.groupby(
    "Gender"
)["Age"].mean()


print("\nAverage Age by Gender:")

print(gender_avg_age)


# ============================================================
# 12. GENDER-WISE AVERAGE SALARY VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

gender_avg_salary.plot(
    kind="bar"
)

plt.xlabel("Gender")

plt.ylabel("Average Salary")

plt.title(
    "Gender-wise Average Salary"
)

plt.xticks(rotation=0)

plt.show()


# ============================================================
# 13. CORRELATION ANALYSIS
# ============================================================

print("\n================ CORRELATION ANALYSIS ================")


# Select numerical columns

numeric_data = df[
    [
        "Age",
        "Experience_Years",
        "Salary"
    ]
]


# Calculate correlation

correlation = numeric_data.corr()


print("\nCorrelation Matrix:")

print(correlation)


# ============================================================
# 14. SALARY AND EXPERIENCE CORRELATION
# ============================================================

experience_salary_corr = df[
    [
        "Experience_Years",
        "Salary"
    ]
].corr()


print("\nExperience vs Salary Correlation:")

print(experience_salary_corr)


# ============================================================
# 15. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8, 6))

plt.imshow(
    correlation,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title(
    "Correlation Matrix"
)

# Display correlation values

for i in range(len(correlation.columns)):

    for j in range(len(correlation.columns)):

        plt.text(
            j,
            i,
            round(
                correlation.iloc[i, j],
                2
            ),
            ha="center",
            va="center"
        )

plt.tight_layout()

plt.show()


# ============================================================
# 16. EXPERIENCE VS SALARY VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Experience_Years"],
    df["Salary"]
)

plt.xlabel(
    "Experience (Years)"
)

plt.ylabel(
    "Salary"
)

plt.title(
    "Experience vs Salary Relationship"
)

plt.show()


# ============================================================
# 17. AGE VS SALARY VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Salary"]
)

plt.xlabel(
    "Age"
)

plt.ylabel(
    "Salary"
)

plt.title(
    "Age vs Salary Relationship"
)

plt.show()


# ============================================================
# 18. SALARY DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df["Salary"],
    bins=20
)

plt.xlabel("Salary")

plt.ylabel("Number of Employees")

plt.title(
    "Salary Distribution"
)

plt.show()


# ============================================================
# 19. FINAL ANALYTICAL SUMMARY
# ============================================================

print("\n====================================================")
print("       EXPERIMENT 4 COMPLETED SUCCESSFULLY")
print("====================================================")

print("\nDataset Information:")

print(
    "Number of Rows:",
    df.shape[0]
)

print(
    "Number of Columns:",
    df.shape[1]
)

print("\nFiltering Performed:")
print("- Salary greater than 50000")
print("- Experience greater than 5 years")
print("- Age greater than 30")

print("\nSorting Performed:")
print("- Salary in descending order")
print("- Experience in descending order")

print("\nStatistical Analysis:")
print(
    "Mean Salary:",
    round(salary_mean, 2)
)

print(
    "Median Salary:",
    round(salary_median, 2)
)

print(
    "Standard Deviation:",
    round(salary_std, 2)
)

print("\nGrouping:")
print("- Average Salary by Gender")
print("- Average Experience by Gender")
print("- Average Age by Gender")

print("\nCorrelation Analysis:")

print(correlation)

print("\nEDA Analysis Completed Successfully!")

print("====================================================")
```
